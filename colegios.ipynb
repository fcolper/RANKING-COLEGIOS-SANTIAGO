```python
import pandas as pd
```

```python
csv_2023 = pd.read_csv('2025-2023.csv', sep=';')
csv_2020 = pd.read_csv('2022-2020.csv', sep=';')
csv_2018 = pd.read_csv('2019-2018.csv', sep=';')
```

```python
csv_2023_rm=csv_2023[csv_2023['CODIGO_REGION']==13.0]
df_filtrado_rm=csv_2023_rm[csv_2023_rm['SITUACION_EGRESO']==1]
df_filtrado_rm['Promedio_len_mat']=(df_filtrado_rm['MATE_ACTUAL']+df_filtrado_rm['CLEC_ACTUAL'])/2
```

```python
df_filtrado_año=df_filtrado_rm[df_filtrado_rm['AÑO']==2023]

df_rm_2023=df_filtrado_año.groupby('RBD', as_index=False).agg({
    'ESTABLECIMIENTO': 'first',
    'AÑO':'mean',
    'SITUACION_EGRESO':'sum',
    'CLEC_ACTUAL':'mean',
    'MATE_ACTUAL':'mean',
    'Promedio_len_mat':'mean',

    })

df_rm_2023['Ranking'] = df_rm_2023['Promedio_len_mat'].rank(method='dense', ascending=False).astype(int)
df_rm_2023['AÑO'].astype('Int64')


df_rm_2023.sort_values('Ranking')
```

```python
df_filtrado_año=df_filtrado_rm[df_filtrado_rm['AÑO']==2024]

df_rm_2024=df_filtrado_año.groupby('RBD', as_index=False).agg({
    'ESTABLECIMIENTO': 'first',
    'AÑO':'mean',
    'SITUACION_EGRESO':'sum',
    'CLEC_ACTUAL':'mean',
    'MATE_ACTUAL':'mean',
    'Promedio_len_mat':'mean',

    })

df_rm_2024['Ranking'] = df_rm_2024['Promedio_len_mat'].rank(method='dense', ascending=False).astype(int)
df_rm_2024['AÑO'].astype('Int64')


df_rm_2024.sort_values('Ranking')
```

```python
df_filtrado_año=df_filtrado_rm[df_filtrado_rm['AÑO']==2025]

df_rm_2025=df_filtrado_año.groupby('RBD', as_index=False).agg({
    'ESTABLECIMIENTO': 'first',
    'AÑO':'mean',
    'SITUACION_EGRESO':'sum',
    'CLEC_ACTUAL':'mean',
    'MATE_ACTUAL':'mean',
    'Promedio_len_mat':'mean',

    })

df_rm_2025['Ranking'] = df_rm_2025['Promedio_len_mat'].rank(method='dense', ascending=False).astype(int)
df_rm_2025['AÑO'].astype('Int64')


df_rm_2025.sort_values('Ranking')
```

```python
df_concat=pd.concat([df_rm_2023,df_rm_2024,df_rm_2025])
df_concat[df_concat['Ranking']==1]
```

```python
# Ordenar el DataFrame por AÑO
df = df_colegios_2018_2025.sort_values(by="AÑO")

# Creamos el gráfico de líneas con plotly express. Cada traza representa un ESTABLECIMIENTO.
fig = px.line(
    df,
    x="AÑO",
    y="Ranking",
    color="ESTABLECIMIENTO",
    markers=True,
    title="Evolución del Ranking de Colegios",
    labels={"Ranking": "Ranking (Escala inversa)", "AÑO": "Año"},
    line_shape="linear",
    hover_data={"Promedio_len_mat": ":.1f", 'ESTABLECIMIENTO': True}
)

# Invertir el eje Y para que el ranking menor (mejor) aparezca arriba
fig.update_yaxes(autorange="reversed")


# Creamos un diccionario que relacione cada establecimiento con su comuna.
# Suponemos que cada establecimiento tiene una única comuna.
estab_to_comuna = df.drop_duplicates("ESTABLECIMIENTO").set_index("ESTABLECIMIENTO")["NOM_COM_RBD"].to_dict()

# Obtenemos la lista de comunas disponibles (sin repetir)
comunas = sorted(df["NOM_COM_RBD"].unique())

# Creamos los botones para el menú desplegable
buttons = []

# Botón para mostrar todos los establecimientos
buttons.append({
    "label": "Todos",
    "method": "update",
    "args": [
        {"visible": [True] * len(fig.data)},
        {"title": "Evolución del Ranking de Colegios"}
    ]
})

# Para cada comuna, definimos la visibilidad de las trazas (cada traza corresponde a un establecimiento)
for comuna in comunas:
    # La visibilidad es True para la traza cuyo establecimiento pertenezca a la comuna actual, False para los demás.
    visibility = [estab_to_comuna.get(trace.name, None) == comuna for trace in fig.data]
    buttons.append({
        "label": comuna,
        "method": "update",
        "args": [
            {"visible": visibility},
            {"title": f"Evolución del Ranking en {comuna}"}
        ]
    })

# Actualizamos la disposición del layout para agregar el menú desplegable a la derecha
fig.update_layout(
    updatemenus=[{
        "buttons": buttons,
        "direction": "down",
        "showactive": True,
        "x": 1,         # Posición horizontal en el 100% (derecha)
        "xanchor": "right",
        "y": 1,         # Posición vertical en el 100% (arriba)
        "yanchor": "top"
    }]
)

fig.show()
```

```python
import plotly.express as px

# Ordenar el DataFrame por AÑO
df = df_colegios_2018_2025.sort_values(by="AÑO")

# Crear un diccionario que relacione RBD con el nombre del establecimiento
rbd_to_estab = df.drop_duplicates("RBD").set_index("RBD")["ESTABLECIMIENTO"].to_dict()

# Crear un diccionario que relacione RBD con la comuna
rbd_to_comuna = df.drop_duplicates("RBD").set_index("RBD")["NOM_COM_RBD"].to_dict()

# Crear el gráfico de líneas, agrupando por RBD pero mostrando el nombre del establecimiento
fig = px.line(
    df,
    x="AÑO",
    y="Ranking",
    color="RBD",  # Se agrupa por RBD
    markers=True,
    title="Evolución del Ranking de Colegios",
    labels={"Ranking": "Ranking (Escala inversa)", "AÑO": "Año"},
    line_shape="linear",
    hover_data={"ESTABLECIMIENTO": True, "Promedio_len_mat": ":.1f"}  # Se muestra el nombre en hover
)

# Invertir el eje Y para que el ranking menor (mejor) aparezca arriba
fig.update_yaxes(autorange="reversed")

# Obtenemos la lista de comunas disponibles (sin repetir)
comunas = sorted(df["NOM_COM_RBD"].unique())

# Creamos los botones para el menú desplegable
buttons = []

# Botón para mostrar todos los establecimientos
buttons.append({
    "label": "Todos",
    "method": "update",
    "args": [
        {"visible": [True] * len(fig.data)},
        {"title": "Evolución del Ranking de Colegios"}
    ]
})

# Para cada comuna, definimos la visibilidad de las trazas (cada traza corresponde a un establecimiento)
for comuna in comunas:
    visibility = [rbd_to_comuna.get(int(trace.name), None) == comuna for trace in fig.data]
    buttons.append({
        "label": comuna,
        "method": "update",
        "args": [
            {"visible": visibility},
            {"title": f"Evolución del Ranking en {comuna}"}
        ]
    })

# Agregar el menú desplegable al layout
fig.update_layout(
    updatemenus=[{
        "buttons": buttons,
        "direction": "down",
        "showactive": True,
        "x": 1,
        "xanchor": "right",
        "y": 1,
        "yanchor": "top"
    }]
)

# **Actualizar la leyenda para que muestre el nombre del establecimiento en lugar del RBD**
fig.for_each_trace(lambda t: t.update(name=f"{rbd_to_estab.get(int(t.name), 'Desconocido')} ({rbd_to_comuna.get(int(t.name), 'Sin Comuna')})"))

fig.show()
```

```python
#Guardar en HTML
fig.write_html("evolucion_ranking_establecimiento.html")
```
